
Series is a 1-dimensional labelled array(single column in a df sense) and  DataFrame is a 2-dimensional table (labelled)which has rows and columns.
operation that can be performed on both are many for example.head() which shows fist few values. operation that can be performed only on a dataframe is for eg, .insert()it adds new column which is possible only for df since series has only column.

```python
# ========== Q2 ==========
```



inplace=True means the operation directly modifies the original Series or DataFrame, instead of just returning a modified result.
Without inplace=True, the original object remains unchanged, and the modified result is returned separately.
Pandas generally discourages its use because it can make code less readable, less convenient for method chaining, and it doesn’t necessarily give better memory efficiency.

```python
# ======== Q3  ========
```


Both are used to handle missing values, but they do it differently.
.fillna() replaces missing values with a suitable value. We use it when the missing data is important and we don't want to lose those records.
Example: If some students have missing marks, we can fill them with the median marks.
.dropna() removes rows or columns containing missing values. We use it when only a few records have missing values and removing them won't affect the analysis much.
Example: If only 5 out of 10,000 customer records have missing data, we can simply remove those 5 rows.

```python
# ======== Q4 ========
```


pd.to_numeric(col) converts the values in a column to numeric type, but raises an error when it goes through a non-numeric value.
With errors='coerce', those invalid values are converted to NaN instead of raising an error.
For real-world messy data, I would use errors='coerce' because the conversion continues even when some entries are invalid, and we can handle those NaN values later.

```python
# ======== Q5 ========
```


.loc selects rows or columns using labels, while .iloc selects them using their integer positions.
For example, df.loc['A'] selects the row whose label is 'A', while df.iloc[0] selects the first row.
They can return different rows when the DataFrame's index labels are not the default 0, 1, 2.... For example, if the index is [10, 20, 30], df.loc[20] returns the row labelled 20, while df.iloc[20] looks for the 21st position and would raise an IndexError if it doesn't exist.

```python
# ======== Q6 ========
```

.map() is used when we want to map existing values to corresponding new values.we can Use it when there is a clear value-to-value mapping.
.apply() is used when we want to perform a function, calculation, or custom operation on the data.  we can Use it when simple mapping or replacement isn't enough.
.replace() is used to replace specific values with other values. we can Use it when we simply need to change certain existing values.

```python
# ======== Q7 ========
```


SettingWithCopyWarning occurs when pandas is not sure whether we are modifying the original DataFrame or a copy, usually because of chained assignment.
For example, instead of modifying filtered data in two steps, use .loc[] directly:
df.loc[df["Marks"] > 80, "Grade"] = "A"
This makes it clear which rows and column we want to modify and avoids the warning.

```python
# ======== Q8 ========
```

pd.cut() divides data into ranges that we define, while pd.qcut() divides the data based on quantiles, so each group has roughly the same number of values.
For annual income, I’d use cut() if I want meaningful categories like Low, Mid, and High based on fixed income ranges. I’d use qcut() if I want the customers divided into groups with roughly equal numbers of customers.

```python
# Prediction:
#    rating
# 0     4.5
# 1     3.8
# 2     4.2
# 3     5.0
#
#    rating
# 0     9.0
# 2     8.4
# 3    10.0
#


import pandas as pd

df = pd.DataFrame({'rating': [4.5, 3.8, 4.2, 5.0]})
df_filtered = df[df['rating'] > 4]
df_filtered['rating'] = df_filtered['rating'] * 2
print(df)
print(df_filtered)
```

```python
import pandas as pd
import numpy as np

df_raw = pd.read_csv('indian_startups_funding.csv')

print("Rows and columns:", df_raw.shape)
print("\nMissing values in each column:")
print(df_raw.isna().sum())
print("\nExact duplicate rows:", df_raw.duplicated().sum())
```

```python
# ========== Q18 ==========
```

```python
# Clean the dataset once here and reuse df for all questions below.

df = df_raw.copy()

df.columns = [
    'startup_name', 'city', 'industry', 'funding_stage',
    'amount', 'funding_date', 'investors', 'founded_year'
]

df['startup_name'] = df['startup_name'].str.strip().str.title()
df['city'] = df['city'].str.strip().str.title()

city_aliases = {
    'Bangalore': 'Bengaluru',
    'Blr': 'Bengaluru',
    'Bombay': 'Mumbai',
    'Bom': 'Mumbai',
    'Gurgaon': 'Gurugram',
    'Ggn': 'Gurugram',
    'Delhi': 'New Delhi',
    'Calcutta': 'Kolkata',
    'Madras': 'Chennai',
    'Hyd': 'Hyderabad'
}
df['city'] = df['city'].replace(city_aliases)

industry_aliases = {
    'Agri Tech': 'AgriTech',
    'Agri-Tech': 'AgriTech',
    'AGRITECH': 'AgriTech',
    'agritech': 'AgriTech',
    'AI ML': 'AI/ML',
    'AI-ML': 'AI/ML',
    'AI/ML': 'AI/ML',
    'AIML': 'AI/ML',
    'Beauty-Tech': 'BeautyTech',
    'BEAUTYTECH': 'BeautyTech',
    'BeautyTech': 'BeautyTech',
    'E-Commerce': 'Ecommerce',
    'E-commerce': 'Ecommerce',
    'ECOMMERCE': 'Ecommerce',
    'ecommerce': 'Ecommerce',
    'Ecommerce': 'Ecommerce',
    'EDTECH': 'EdTech',
    'Ed-Tech': 'EdTech',
    'EdTech': 'EdTech',
    'Edtech': 'EdTech',
    'EV': 'EV',
    'ev': 'EV',
    'Electric Vehicles': 'EV',
    'Electric-Vehicles': 'EV',
    'FINTECH': 'FinTech',
    'Fin-Tech': 'FinTech',
    'FinTech': 'FinTech',
    'Fintech': 'FinTech',
    'fintech': 'FinTech',
    'FOOD-TECH': 'FoodTech',
    'Food Tech': 'FoodTech',
    'Food-Tech': 'FoodTech',
    'FoodTech': 'FoodTech',
    'foodtech': 'FoodTech',
    'FT': 'FoodTech',
    'GAMING': 'Gaming',
    'Games': 'Gaming',
    'Gaming': 'Gaming',
    'gaming': 'Gaming',
    'HEALTHTECH': 'HealthTech',
    'Health Tech': 'HealthTech',
    'Health-Tech': 'HealthTech',
    'HealthTech': 'HealthTech',
    'healthtech': 'HealthTech',
    'LOGISTICS': 'Logistics',
    'Logistics': 'Logistics',
    'logistics': 'Logistics',
    'PROPTECH': 'PropTech',
    'Prop-Tech': 'PropTech',
    'PropTech': 'PropTech',
    'SAAS': 'SaaS',
    'SaaS': 'SaaS',
    'Saas': 'SaaS',
    'saas': 'SaaS',
    'TRAVELTECH': 'TravelTech',
    'Travel-Tech': 'TravelTech',
    'TravelTech': 'TravelTech'
}
df['industry'] = df['industry'].replace(industry_aliases)

funding_stage_aliases = {
    'BRIDGE': 'Bridge', 'Bridge': 'Bridge', 'bridge': 'Bridge',
    'IPO': 'IPO', 'ipo': 'IPO',
    'PRE-SEED': 'Pre-Seed', 'Pre Seed': 'Pre-Seed',
    'Preseed': 'Pre-Seed', 'Pre-Seed': 'Pre-Seed',
    'PRE-SERIES A': 'Pre-Series A', 'Pre Series A': 'Pre-Series A',
    'Pre-Series A': 'Pre-Series A', 'Pre-A': 'Pre-Series A',
    'Pre-IPO': 'Pre-IPO',
    'SEED': 'Seed', 'Seed': 'Seed', 'seed': 'Seed',
    'SERIES A': 'Series A', 'Series A': 'Series A',
    'Series-A': 'Series A', 'series a': 'Series A', 'series_a': 'Series A',
    'SERIES B': 'Series B', 'Series B': 'Series B',
    'Series-B': 'Series B', 'series b': 'Series B', 'series_b': 'Series B',
    'SERIES C': 'Series C', 'Series C': 'Series C',
    'Series-C': 'Series C', 'series c': 'Series C', 'series_c': 'Series C',
    'SERIES D': 'Series D', 'Series D': 'Series D',
    'Series-D': 'Series D', 'series d': 'Series D', 'series_d': 'Series D',
    'SERIES E': 'Series E', 'Series E': 'Series E',
    'Series-E': 'Series E', 'series e': 'Series E', 'series_e': 'Series E'
}
df['funding_stage'] = df['funding_stage'].replace(funding_stage_aliases)

df['amount'] = (
    df['amount'].astype(str)
    .str.replace('₹', '', regex=False)
    .str.replace('INR', '', regex=False)
    .str.replace('Crore', '', regex=False)
    .str.replace('Cr', '', regex=False)
    .str.replace(',', '', regex=False)
    .str.strip()
    .replace('Undisclosed', np.nan)
)
df['amount'] = pd.to_numeric(df['amount'], errors='coerce')

df['funding_date'] = pd.to_datetime(
    df['funding_date'], errors='coerce', dayfirst=True
)

df = df.drop_duplicates().copy()

print(sorted(df.columns))
```

The raw file has been cleaned once here: column names are standardised, text and aliases are cleaned, funding amounts and dates are converted, and exact duplicate rows are removed. The same cleaned `df` is used from Q19 onward.
